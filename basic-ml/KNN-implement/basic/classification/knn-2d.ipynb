```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("./data/iris_2D.csv")
data
```

```python
# Get x
x_data = data[["Petal_Length", "Petal_Width"]].to_numpy()
x_data.reshape(6, 2)
x_data
```

```python
# Get y
y_data = data["Label"].to_numpy()
y_data
```

```python
x_test = np.array([[2.4, 0.8]])
x_test
```

IMPLEMENT SIMPLE KNN 2D

```python
# COMPUTE DISTANCES
distances = np.sqrt(np.sum(np.square(x_data - x_test), axis=1))
print(distances)
```

```python
# Get the top k
top_k = np.argsort(distances)
print(f"Index of sorting nearest neighbors: {top_k}")

k = 3
top_k = top_k[:k]
print(f"Index of top {k} nearest neighbors: {top_k}")
```

```python
# Get corresponding y values
values_top_k = y_data[top_k]
print(f"Values of top {k} nearest neighbors: {values_top_k}")
```

```python
# Get majortiy
num_zeros = np.sum(values_top_k == 0)
num_ones = np.sum(values_top_k == 1)
print(f"Number of zeros: {num_zeros}")
print(f"Number of ones: {num_ones}")

# classification
if num_zeros > num_ones:
    print("The class is 0")
else:
    print("The class is 1")
```

IMPLEMENT USING LIB

```python
from sklearn.neighbors import KNeighborsClassifier

classifier = KNeighborsClassifier(n_neighbors=3)
classifier.fit(x_data, y_data)

y_pred = classifier.predict(x_test)
print(y_pred)
```
