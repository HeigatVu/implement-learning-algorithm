```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import neighbors, datasets
import pandas as pd
```

NORMALIZATION IN CM

```python
data = pd.read_csv("./data/iris_2D.csv")
data
```

```python
# Get x
x_data = data[["Petal_Length", "Petal_Width"]].to_numpy()
x_data

# Get y
y_data = data["Label"].to_numpy()
y_data
```

```python
x_test = [[2.6, 0.7]]
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
classifier = KNeighborsClassifier(n_neighbors=6)
classifier.fit(x_data, y_data)
```

```python
y_pred_pred = classifier.kneighbors(x_test)
y_pred_pred
```

```python
y_pred = classifier.predict(x_test)
print(y_pred)
```

CM

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
data_cm = pd.read_csv('./data/iris_2D.csv')

# get x
x_data_cm = data_cm[['Petal_Length', 'Petal_Width']].to_numpy()
x_data_cm = x_data_cm.reshape(6, 2)
print(x_data_cm)

# get y
y_data_cm = data_cm['Label'].to_numpy()
print(y_data_cm)
```

```python
scaler_cm = StandardScaler()
scaler_cm.fit(x_data_cm)

x_data_cm = scaler_cm.transform(x_data_cm)
print(x_data_cm)
```

```python
classifier_cm_scaler = KNeighborsClassifier(n_neighbors=6)
classifier_cm_scaler.fit(x_data_cm, y_data_cm)

# unknown input
x_test_cm = [[2.6, 0.7]]
x_test_cm = scaler_cm.transform(x_test_cm)
print(x_test_cm)

y_pred_cm = classifier_cm_scaler.kneighbors(x_test_cm)
print(y_pred_cm)
```
