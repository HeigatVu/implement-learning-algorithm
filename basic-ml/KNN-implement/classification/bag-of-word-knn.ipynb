```python
import sklearn
from sklearn.neighbors import KNeighborsClassifier
from sklearn.feature_extraction.text import CountVectorizer
import numpy as np
```

```python
vectorizer = CountVectorizer()
corpus = ["góp gió gặt bão",
          "có làm mới có ăn",
          "đất lành chim đậu",
          "ăn cháo đá bát",
          "gậy ông đập lưng ông",
          "qua cầu rút ván"]
x_data = vectorizer.fit_transform(corpus)
x_data = x_data.toarray()
```

```python
print(x_data)
print(x_data.shape)
```

```python
y_data = np.array([1, 1, 1, 0, 0, 0])
y_data
```

```python
x_test = vectorizer.transform(['không làm cạp đất mà ăn']).toarray()
x_test
```

```python
# Compute distances
distances = np.sqrt(np.sum((x_data - x_test)**2, axis=1))
print(distances)
```

```python
# Get top k
top_k = np.argsort(distances)
print(top_k)

k = 3
top_k = top_k[:k]
print(top_k)
```

```python
value_y_data = y_data[top_k]
print(value_y_data)

number_of_ones = np.sum(value_y_data == 1)
number_of_zeros = np.sum(value_y_data == 0)

if number_of_ones > number_of_zeros:
    print("1")
else:
    print("0")
```

USING LIB

```python
classifier = KNeighborsClassifier(n_neighbors=3)
classifier.fit(x_data, y_data)

y_pred = classifier.predict(x_test)
print(y_pred)
```
