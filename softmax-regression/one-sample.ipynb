```python
import numpy as np
```

```python
def convert_one_hot(y, k):
    one_hot = np.zeros((len(y), k), dtype=int)
    one_hot[np.arange(len(y)), y] = 1
    return one_hot
```

```python
y = np.array([0, 2, 1])
n_classes = 3
one_hot_y = convert_one_hot(y, n_classes)
```
