```python
import torch
```

```python
x1 = torch.tensor([1, 2, 3], dtype=torch.float32)
x2 = torch.tensor([4, 5, 6], dtype=torch.float32)

# Forward layer
h0f = torch.tensor([0, 0], dtype=torch.float32)

wxh1f = torch.tensor([[-1, 0, 0.1], [1, 0, -0.1]], dtype=torch.float32)
wxh2f = torch.tensor([[-1, 0, 0.1], [1, 0, -0.1]], dtype=torch.float32)
bxh1f = 1.0
bxh2f = 1.0

whh0f = torch.tensor([0, -1], dtype=torch.float32)
whh1f = torch.tensor([0, -1], dtype=torch.float32)
bhh0f = 1.0
bhh1f = 1.0

# Backward layer
h0b = torch.tensor([0, 0], dtype=torch.float32)

wxh1b = torch.tensor([[-1, 0, 0.1], [1, 0, -0.1]], dtype=torch.float32)
wxh2b = torch.tensor([[-1, 0, 0.1], [1, 0, -0.1]], dtype=torch.float32)
bxh1b = 1.0
bxh2b = 1.0

whh0b = torch.tensor([0, -1], dtype=torch.float32)
whh1b = torch.tensor([0, -1], dtype=torch.float32)
bhh0b = 1.0
bhh1b = 1.0

who2 = torch.tensor([1, 1, 1, 1], dtype=torch.float32)
bho2 = 0.0
```

```python
# forward pass with different inputs
r_wxh1f = torch.matmul(wxh1f, x1) + bxh1f
r_wxh2f = torch.matmul(wxh2f, x2) + bxh2f
r_whh0f = torch.matmul(whh0f, h0f) + bhh0f
r_whh1f = torch.matmul(whh1f, h0f) + bhh1f
h1f = torch.tanh(r_wxh1f + r_whh0f)
h2f = torch.tanh(r_wxh2f + r_whh1f)

# backward pass with different inputs
r_wxh1b = torch.matmul(wxh1b, x2) + bxh1b
r_wxh2b = torch.matmul(wxh2b, x1) + bxh2b
r_whh0b = torch.matmul(whh0b, h0b) + bhh0b
r_whh1b = torch.matmul(whh1b, h0b) + bhh1b
h1b = torch.tanh(r_wxh1b + r_whh0b)
h2b = torch.tanh(r_wxh2b + r_whh1b)

output = torch.cat((h2f, h2b), dim=0)
result = torch.matmul(who2, output) + bho2
print("Output:", result.item())
```
