```python
import numpy as np
```

```python
X = np.array([
    [0.5],
    [1.0],
    [1.5],
    [2.0],
    [2.5],
    [3.0],
    [3.5],
    [4.0],
])

Y = np.array([0, 0, 0, 1, 0, 0, 1, 1, 1])
```

```python
def convert_one_hot(y, k):
    one_hot = np.zeros((len(y), k))
    one_hot[np.arange(len(y)), y] = 1
    return one_hot

n_classes = 2
Y_one_hot = convert_one_hot(Y, n_classes)
print(Y_one_hot)
```

```python
X = np.hstack([np.ones((X.shape[0], 1)), X])
print(X)
```

```python
x = X[0].reshape((2, 1))
print(x)
y = Y_one_hot[0].reshape((2, 1))
print(y)
```

```python
theta = np.array([[0.1, 0.2], [0.3, 0.4]])
print(theta)
```

```python
def softmax_fn(z):
    return np.exp(z) / np.sum(np.exp(z))

def predict(x, theta):
    z = theta.T.dot(x)
    y_hat = softmax_fn(z)
    return z, y_hat

z, y_hat = predict(x, theta)
print(z)
print(y_hat)
```

```python
def compute_loss(y_hat, y):
    loss = -np.log(y.T.dot(y_hat))
    return loss

loss = compute_loss(y_hat, y)
loss
```

```python
def compute_gradient(y_hat, y, x):
    gradient = x.dot((y_hat - y).T)
    return gradient

gradient = compute_gradient(y_hat, y, x)
gradient
```

```python
learning_rate = 0.01

def update_weight(theta, gradient, learning_rate):
    theta -= (learning_rate * gradient)
    return theta

theta = update_weight(theta, gradient, learning_rate)
theta
```
